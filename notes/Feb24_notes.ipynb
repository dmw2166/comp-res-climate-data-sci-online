### SciPy

```python
from scipy import stats
```

```python
dir(stats)
```

```python
# Load columns 3 (DATE) and 6 (TMIN) from the CSV as strings, skipping the header row
# filling_values='' fills any missing cells with an empty string instead of crashing
data = np.genfromtxt('/data9/G5053/data/shr/centralpark.csv', 
                     delimiter=',', dtype=str, skip_header=1,
                     usecols=(3, 6), filling_values='')

# The dates come in as '"1869-01-01"' — strip the quote characters from every element
dates = np.char.strip(data[:, 0], '"')

# Same for tmin values — '"19"' becomes '19'
tmin  = np.char.strip(data[:, 1], '"')

# Some rows have no TMIN measurement, leaving an empty string ''
# Replace those with 'nan' so we can convert to float (float('nan') works, float('') doesn't)
tmin[tmin == ''] = 'nan'

# Boolean mask: True for every row where the date starts with '1869'
mask1869 = np.array([d.startswith('1869') for d in dates])
mask2023 = np.array([d.startswith('2023') for d in dates])

# Apply the mask to grab only 1869/2023 rows, then convert from strings to numbers
t1869 = tmin[mask1869].astype(float)
t2023 = tmin[mask2023].astype(float)
```

```python
t1869
```

```python
t2023
```

```python
plt.plot(t1869, label='1869')
plt.plot(t2023, label='2023')
plt.legend()
plt.xlabel('Day of year')
plt.ylabel('Minimum daily temperature (deg F)')
```

```python
t_mean1869 = np.nanmean(t1869)
t_mean1869
```

```python
t_mean2023 = np.nanmean(t2023)
t_mean2023
```

Null hypothesis is that the mean minimum daily T in 1869 is equal to the mean minimum daily mean in 2023

```python
t_val, p_val = stats.ttest_rel(t1869,t2023,nan_policy='omit')
p_val
```

null hypothesis that the difference on average in daily min temps between 2023 and 1869 is 2 degrees F

```python
t_val, p_val = stats.ttest_1samp(t2023-t1869, 2, nan_policy='omit')
p_val
```

```python
t_val, p_val = stats.ttest_1samp(t2023-t1869, 3.5, nan_policy='omit')
p_val
```
