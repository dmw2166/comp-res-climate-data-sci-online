```python
def say_hello_to(name):
    """ Return a greeting to a name. """
    return 'Hello ' + str(name) + '!'
```

```python
say_hello_to('Dan')
```

```python
say_hello_to('World')
```

```python
say_hello_to(10)
```

```python
def say_hello_or_bjr(name, french=False):
    """ function to say a greeting in english or french """ 
    if french: 
        greeting = 'Bonjour '
    else: 
        greeting = 'Hello '
    return greeting + name
```

```python
say_hello_or_bjr('Dan')
```

```python
say_hello_or_bjr('Dan', french=True)
```

```python
def fib(n):
    """ print a fibonacci series with n elements"""
    l = [0,1]
    for i in range(n-2): 
        l.append(l[-1] + l[-2])
    return l 
```
